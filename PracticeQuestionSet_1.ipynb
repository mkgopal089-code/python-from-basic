Write a Python program that takes three integers from the user and prints the
largest number. If all three numbers are equal, print "All are Equal"



```python
num1 = int(input("Enter number 1st:"))
num2 = int(input("Enter number 2nd:"))
num3 = int(input("Enter number 3rd:"))

if num1 == num2 == num3:
  print("All are Equal")
elif num1>=num2 and num1>=num3:
  print(num1)
elif num2>=num1 and num2>=num3:
  print(num2)
else:
  print(num3)
```

```python
def add(numbers):


  total = 0
  for n in numbers:
    total = total + n
  return total

print(add(20,30))
```

```python
def add(*numbers):

  #{*numbers} allows the function to accept any number of positional arguments:


  total = 0
  for n in numbers:
    total = total + n
  return total

print(add(20,30))
```
