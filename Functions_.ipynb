map, filter and reduce functions

Syntax --- map(function, iterable)

```python
numbers = [12,43,17,76,97,57]
even_odd = lambda n : "Even" if n%2 == 0 else "Odd"
```

```python
even_odd = list(map(lambda n : "Even" if n%2 == 0 else "Odd",numbers))
print(even_odd)
```

```python
words = ["Amazing","Pipeline","Python","Springboot","Attendance","Proxy","is","why","wow!","class","Tech"]

word_length = lambda word: len(word)
```

```python
word_length("Spectacular")
```

```python
word_length = list(map(lambda word: len(word),words))
print(word_length)
```

Filter - Applies a filtering condition on Iterable

```python
## Select all the even numbers from a list of numbers

num = [12,32,45,56,76,88,19,13,94]

Odd = list(filter(lambda n: n%2 != 0,num))
print(Odd)
```

```python
## In words filter out all the words with no. of Charecter more than 5

Len_words = list(filter(lambda word: len(word)> 5,words))
print(Len_words)
```

Reduce - Collapses a sequence on a single value; like aggregation

Syntax - reduce(function, iterable, [initial_value])

```python
numbers = [12,43,17,76,97,57]
```

```python
from functools import reduce
Total = reduce(lambda total,n: total+n,numbers)
print(Total)
```
