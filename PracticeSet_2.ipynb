Student Performance Analyzer
Write a Python program that accepts marks of n students.
The program should:
1. Store all marks in a list.
2. Calculate:
    ○ Average marks
    ○ Highest marks
    ○ Lowest marks
3. Display the grade:
Average >= 90 : Excellent
Average >= 75 : Good
Average >= 50 : Average
Below 50 : Need Improvement

```python
n =int(input("enter no of students:"))

marks = []

for i in range(n):
  marks.append(int(input()))


average = sum(marks) /n

print("Average" ,average)
print("Highest",max(marks))
print("lowest",min(marks))


if(average>=90):
  print("Grade: Excellent")


elif(average>=75):
  print("Grade: Good")


elif(average>=50):
  print("Grade: Average")


else:
  print("Grade: Need Improvement")
```

Dictionary Based Problem
A company stores employee salaries:
employees = {
"Amit":50000,
"Rahul":60000,
"Neha":75000,
"Priya":45000
}

Write a program to print employees whose salary is greater than 50000.
Expected Output:
Rahul 60000
Neha 75000

```python
employees ={

"Amit":50000, "Rahul":60000, "Neha":75000,"Priya":45000
}

for name,salary in employees.items():

  if salary > 50000:
    print(name,salary)
```

**Count_vowels**

```python
def count_vowels(text):

  count = 0

  for ch in text:
    if ch.lower() in "aeiou":
      count +=1
  return count

print(count_vowels("Harsh Gopal"))
```

Create a function:
calculate_average()

which accepts any number of values and returns their average.
Example:

print(calculate_average(10,20,30))

print(calculate_average(5,10,15,20,25))

```python
def calculate_average(*numbers):
  total = 0

  for n in numbers:
    total = total + n
  return total / len(numbers)

print(calculate_average(10,20,30))
print(calculate_average(5,10,15,20,25))
```

Recursion Problem

Write a recursive function to calculate the sum of first n natural numbers.

```python
def sum_number(n):
  if n ==0:
    return 0

  return n + sum_number(n-1)

n = int(input())

print(sum_number(n))
```

Given:
students={
"Aman":[80,90,70],
"Riya":[60,75,85],
"Karan":[90,95,100]
}

Write a program that:
1. Calculates average marks of every student.

2. Stores students having average above 80 into a new dictionary.

3. Display the new dictionary.

Expected Output:
{
'Aman':80.0,
'Karan':95.0

```python
students={ "Aman":[80,90,70], "Riya":[60,75,85], "Karan":[90,95,100] }

result ={}

for name,marks in students.items():
  avg = sum(marks)/len(marks)
  if avg > 80:
    result[name] =avg

print(result)
```

Lambda + Filter Problem
Given:

numbers=[12,15,20,25,30,35,40]
Using filter() and lambda, print only numbers which are:

● Even

● Greater than 20

Expected Output:
[30,40]

```python
numbers=[12,15,20,25,30,35,40]

result = list(
    filter(
        #lambda arguments: expression
        lambda x : x%2 ==0 and x>20,
        numbers
    )
)
print(result)
```
