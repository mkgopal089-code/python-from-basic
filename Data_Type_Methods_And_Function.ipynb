# Prime Number Check

```python
num = int(input("Enter a number: "))
if num <= 0:
  print("Invalid Input. Input a Natural Number")
elif num == 1:
  print("The Number is Not Prime")
else:
  for i in range(2,num):
    if num % i == 0:
      print("The Number is Not Prime")
      break
  else:
    print("The Number is Prime")
```

| `title()`      | Capitalizes the first letter of every word                               | `"python programming".title()` → `"Python Programming"`      |
| `capitalize()` | Capitalizes only the first character                                     | `"python programming".capitalize()` → `"Python programming"` |
| `swapcase()`   | Converts uppercase to lowercase and vice versa                           | `"PyThOn".swapcase()` → `"pYtHoN"`                           |

# **Modification**

| Method           | Purpose                                                                 | Example                                                       |

| `replace()`      | Replaces one substring with another                                     | `"I like Java".replace("Java","Python")` → `"I like Python"`  |

| `split()`        | Splits a string into a list                                             | `"Python SQL Excel".split()` → `['Python', 'SQL', 'Excel']`   |

| `splitlines()`   | Splits a multiline string into a list of lines                          | `"A\nB\nC".splitlines()` → `['A', 'B', 'C']`                  |

| `partition()`    | Splits a string into three parts at the first occurrence of a separator | `"apple-orange".partition("-")` → `('apple', '-', 'orange')`  |
| `rpartition()`   | Splits a string into three parts at the last occurrence of a separator  | `"a-b-c".rpartition("-")` → `('a-b', '-', 'c')`               |
| `center()`       | Centers the string within a specified width                             | `"Python".center(12,"*")` → `"***Python***"`                  |

| `zfill()`        | Pads a numeric string with leading zeros                                | `"45".zfill(5)` → `"00045"`                                   |
| `expandtabs()`   | Replaces tab characters with spaces                                     | `"A\tB".expandtabs(4)` → `"A   B"` *(tab expanded to spaces)* |
| `translate()`    | Replaces characters using a translation table                           | `"cat".translate(str.maketrans("cat","dog"))` → `"dog"`       |
| `maketrans()`    | Creates a translation table for use with `translate()`                  | `str.maketrans("abc","123")`                                  |
| `removeprefix()` | Removes a specified prefix if present                                   | `"unhappy".removeprefix("un")` → `"happy"`                    |
| `removesuffix()` | Removes a specified suffix if present                                   | `"reading".removesuffix("ing")` → `"read"`                    |
