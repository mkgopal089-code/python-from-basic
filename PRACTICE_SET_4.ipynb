# Write a function count_positive(*args) that accepts any number of numeric arguments and returns how many of them are positive (strictly greater than 0).

```python
def count_positive(*args):
 count= 0
 for num in args:
  if num > 0:
    count +=1
 return count

print(count_positive(1, -2, 3, 0, -4, 5))

```

# Write a recursive function sum_of_digits(n) that returns the sum of the digits of a non-negative integer n.

```python
def sum_of_digits(n):
  if n == 0:
    return 0
  elif n < 10:
    return n
  else:
    return int(n % 10) + sum_of_digits(int(n / 10))

print(sum_of_digits(345))
```

# Write a function calculate_average(*args) that accepts any number of numeric arguments and returns their average,without using the built-in sum() or len() functions.

```python
def calculate_average(*args):
  total=0
  count=0

  for num in args:
    total = total + num
    count =count + 1

  return total/count

print(calculate_average(1, 2, 3, 4, 5))
```

# Write a function classify_numbers(*args) that accepts a variable number of numeric arguments and returns a tuple containing the count of positive numbers, negative numbers, and zeros among them, in that order.

```python
def classify_numbers(*args):
  positive_count =0
  negative_count=0
  zero_count=0


  for num in args:
    if num > 0:
      positive_count +=1
    elif num <0:
      negative_count +=1
    else:
      zero_count +=1
  return positive_count, negative_count, zero_count

print(classify_numbers(1, -2, 3, 0, -4, 5))


```

# Write a program that takes a list of numbers and: (1) filters out only the numbers divisible by 3 using filter() with a lambda, (2) cubes each of those numbers using map() with a lambda, and (3) adds up all the cubed values using reduce() from the functools module. Return the final sum. If no numbers are divisible by 3, return 0.

```python
from functools import reduce

nums = [1,2,3,4,5,6,7,8]

div_by_3 = list(filter(lambda x: x % 3 ==0,nums))

cubes = list(map(lambda x: x**3,div_by_3))

if cubes:
  result = reduce(lambda x,y : x+y,cubes)
else:
  result =0

print(result)
```

Another way taking as a **input**

```python
from functools import reduce

n = int(input("Enter number of elements: "))

nums = []

for i in range(n):
    num = int(input("Enter number: "))
    nums.append(num)

# Step 1: Filter numbers divisible by 3
div_by_3 = list(filter(lambda x: x % 3 == 0, nums))

# Step 2: Cube them
cubes = list(map(lambda x: x**3, div_by_3))

# Step 3: Sum using reduce
if cubes:
    result = reduce(lambda a, b: a + b, cubes)
else:
    result = 0

print("Sum of cubes =", result)
```
