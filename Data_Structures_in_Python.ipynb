# **Tuple Operations**

| Utility / Operation                | Operator / Syntax     | Example                 | Result                 |
| ---------------------------------- | --------------------- | ----------------------- | ---------------------- |
| **Create tuple**                   | `()`                  | `t = (10, 20, 30)`      | `(10, 20, 30)`         |
| **Single-element tuple**           | `(value,)`            | `t = (10,)`             | `(10,)`                |
| **Access element**                 | `[]`                  | `t[2]`                  | `30`                   |
| **Negative indexing**              | `[-1]`                | `t[-1]`                 | `50`                   |
| **Access range**                   | `[start:end]`         | `t[1:4]`                | `(20, 30, 40)`         |
| **Slice from beginning**           | `[:end]`              | `t[:3]`                 | `(10, 20, 30)`         |
| **Slice to end**                   | `[start:]`            | `t[2:]`                 | `(30, 40, 50)`         |
| **Reverse tuple**                  | `[::-1]`              | `t[::-1]`               | `(50, 40, 30, 20, 10)` |
| **Skip elements**                  | `[::step]`            | `t[::2]`                | `(10, 30, 50)`         |
| **Concatenate tuples**             | `+`                   | `(1,2) + (3,4)`         | `(1,2,3,4)`            |
| **Repeat tuple**                   | `*`                   | `(1,2) * 3`             | `(1,2,1,2,1,2)`        |
| **Check membership**               | `in`                  | `20 in t`               | `True`                 |
| **Check absence**                  | `not in`              | `60 not in t`           | `True`                 |
| **Compare tuples**                 | `==`, `!=`            | `(1,2) == (1,2)`        | `True`                 |
| **Greater than**                   | `>`                   | `(3,4) > (1,5)`         | `True`                 |
| **Less than**                      | `<`                   | `(1,2) < (1,5)`         | `True`                 |
| **Unpack tuple**                   | Assignment            | `a,b,c = (1,2,3)`       | `a=1,b=2,c=3`          |
| **Extended unpacking**             | `*`                   | `a,*b = (1,2,3,4)`      | `a=1,b=[2,3,4]`        |
| **Nested access**                  | `[][]`                | `t[1][0]`               | Depends on nested data |
| **Loop through tuple**             | `for`                 | `for x in t:`           | Iterates elements      |
| **Tuple comprehension equivalent** | `tuple()` + generator | `tuple(x*2 for x in t)` | `(20,40,60,80,100)`    |
| **Convert to list**                | `list()`              | `list(t)`               | `[10,20,30,40,50]`     |
| **Convert from list**              | `tuple()`             | `tuple([1,2,3])`        | `(1,2,3)`              |
| **Sort values**                    | `sorted()`            | `sorted(t)`             | `[10,20,30,40,50]`     |
| **Reverse**                        | `reversed()`          | `tuple(reversed(t))`    | `(50,40,30,20,10)`     |

| Category                 | Operator | Example      | Result          | Meaning                                     |

**Dictionary Functions and Methods**

| Method / Function    | Syntax                   | What it does                         | Example                              |
| -------------------- | ------------------------ | ------------------------------------ | ------------------------------------ |
| Access using key     | `dict[key]`              | Returns value for a key              | `student_scores["Rohan"]` → `91`     |
| `get()`              | `dict.get(key)`          | Safely retrieves value               | `student_scores.get("Rohan")` → `91` |
| `get()` with default | `dict.get(key, default)` | Returns default if key doesn't exist | `student_scores.get("Raj", 0)` → `0` |
| `keys()`             | `dict.keys()`            | Returns all keys                     | `student_scores.keys()`              |
| `values()`           | `dict.values()`          | Returns all values                   | `student_scores.values()`            |
| `items()`            | `dict.items()`           | Returns key-value pairs              | `student_scores.items()`             |


**Adding and Updating Data in Dictionary**

| Method / Approach | Syntax                     | Purpose                                  |
| ----------------- | -------------------------- | ---------------------------------------- |
| Direct assignment | `d[key] = value`           | Add a new key or update existing key     |
| `update()`        | `d.update({...})`          | Add/update multiple key-value pairs      |
| `setdefault()`    | `d.setdefault(key, value)` | Add key only if it doesn't already exist |


**Removing Data**

| Method               | Syntax                | Purpose                                          | If Key Doesn't Exist |
| -------------------- | --------------------- | ------------------------------------------------ | -------------------- |
| `pop()`              | `d.pop(key)`          | Removes specified key and returns its value      | `KeyError`           |
| `pop()` with default | `d.pop(key, default)` | Removes key safely                               | Returns default      |
| `popitem()`          | `d.popitem()`         | Removes and returns last inserted key-value pair | `KeyError` if empty  |
| `del`                | `del d[key]`          | Deletes specified key                            | `KeyError`           |
| `clear()`            | `d.clear()`           | Removes everything                               | —                    |


**Checking and Searching**

| Operation            | Syntax                     | Example                     | Result |
| -------------------- | -------------------------- | --------------------------- | ------ |
| Check key            | `key in d`                 | `"Rohan" in d`              | `True` |
| Check key absence    | `key not in d`             | `"Raj" not in d`            | `True` |
| Check value          | `value in d.values()`      | `91 in d.values()`          | `True` |
| Check key-value pair | `(key,value) in d.items()` | `("Rohan",91) in d.items()` | `True` |
